```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv("CapitalPunishment.csv")
```

```python
df.head(10)
```

```python
df_new = df.copy()
```

```python
data = {
    'Region and Jurisdiction': ['U.S. total','Northeast', 'Midwest', 'South', 'West'],
    'Prisoners Under Sentence of Death, 12/31/21 (White)': [1013, 46, 84, 511, 353],
    'Prisoners Under Sentence of Death, 12/31/21 (Black)': [949, 49, 86, 511, 286]
}

df_new = pd.DataFrame(data)

df_new['Total Prisoners'] = df_new['Prisoners Under Sentence of Death, 12/31/21 (White)'] + df_new['Prisoners Under Sentence of Death, 12/31/21 (Black)']
df_new = df_new.sort_values(by='Total Prisoners', ascending=False)


plt.figure(figsize=(12, 6))

sns.barplot(x='Region and Jurisdiction', 
            y='Prisoners Under Sentence of Death, 12/31/21 (White)', 
            data=df_new, 
            color='pink', 
            label='White', 
            dodge=True)  

sns.barplot(x='Region and Jurisdiction', 
            y='Prisoners Under Sentence of Death, 12/31/21 (Black)', 
            data=df_new, 
            color='Blue', 
            label='Black', 
            dodge=True)  

plt.title('Comparison of White and Black Prisoners Under Sentence of Death (12/31/21) by Region')
plt.xlabel('Region and Jurisdiction')
plt.ylabel('Number of Prisoners Under Sentence of Death')

plt.xticks(rotation=45)

plt.tight_layout()

plt.legend(title="Race")
plt.show()
```
